## Step 4: Work with the DataFrame



### 4a. FILTERING: We will filter the dataset

`DFname[]`: `[]` are used for indexing, or in other words to select parts of the dataset.

We can specify a condition inside the `[]` to limt our results to only those rows that satisfy the condition.

```python
#remove the Toyota Corolla row
df2 = df[df.index != 'Toyota Corolla']
print(df2)
```

### 4b. RESHAPING: We can select only some columns

In Pandas, when we want to selct more than one column, we use `[]` and separate the col names with a comma.





```python
#create a new dataframe with the first two columns
df3 = df2 [ ['mpg', 'cyl'] ]
print(df3)
```

### 4c. SORTING the rows

The command `sort_values()` can be used to sort based on the columns specified.

if you want to sort in reverse, use another parameter `ascending`. You have to provide it `True` or `False` values for each column you sort on. Default is `True` for all columns.


```python
#sort the rows by mpg
df3.sort_values(by=['mpg'], ascending = False)
print(df3)
```

### 4d. AGGREGATIONS

We can perform aggregations using the `groupby()` method or command on a Dataframe. This will allow us to compute  statistics within groups in our data.

`Df.groupby('colname')` or `DF.groupby(['col1, 'col2']))` if there are multiple columns that identify the groups.

After that we can call the `size()` method to get the number of observations (rows) in each group.

Finally, we will rename this new column as 'count' and reset the index of our output DataFrame.


```python
#aggregate the data to get the number of cars with each cylinder count
df4 = df3.groupby('cyl').size().reset_index(name='count')
print(df4)
```
