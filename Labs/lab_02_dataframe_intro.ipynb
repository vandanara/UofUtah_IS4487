## **Part 2: Getting data into pandas**

In this case we will load data from the statsmodels.org library

MTCARS is a traditional dataset used for learning.



```python
mtcars = sm.datasets.get_rdataset("mtcars", "datasets", cache=True).data
df = pd.DataFrame(mtcars)
df['carmodel'] = df.index
print(df)
```

## **Part 3: Preview the Dataset**

When you first open a new dataset in Python, tools like info(), head(), and describe() help you quickly understand what you’re working with.

- `df.info()` gives you a high-level overview of the dataset. It tells you how many rows and columns there are, the names of the columns, the data types (like numbers vs. text), and whether there are any missing values. You should look here to spot issues like unexpected data types or columns with lots of missing data.
- `df.head()` shows you the first few rows of the dataset. This lets you actually see what the data looks like—how values are formatted, what kind of information each column contains, and whether anything looks strange or inconsistent.
- `df.describe()` provides summary statistics for numerical columns, like the average, minimum, maximum, and quartiles. You should use this to get a sense of the range of values, typical values, and whether there might be outliers or unusual data points.

Together, these three tools give you a quick “first impression” of your dataset so you can start asking better questions and avoid mistakes early on.
